Функция для чистки названий компании:

```python
import re

def norm(x):
    if pd.isna(x):
        return None
    x = str(x).lower().strip()
    x = re.sub(r'["'']', '', x)
    x = re.sub(r'[()\\[\\],.:;]', ' ', x)
    x = re.sub(r'[-_/]', ' ', x)
    x = re.sub(r'\s+', ' ', x).strip()
    return x
```

Теперь такой проблемы не наблюдается - видно, что количество совпавших тайтлов из companies снизилось

```python
df_merge_companies = c.merge(
    v,
    on='company_name',
    how='left',
    suffixes=('_company', '_video')
)

df_merge_companies.head()
```

```python
df_merge_all = c.merge(
    v,
    on='company_name',
    how='outer',
    suffixes=('_company', '_video')
)

df_merge_all.head()
```

Остались компании, которые не попали в итоговый смердженный датасет

Судя по просмотренным в экселе значениям этих компаний и правда нет в данных, собранных с DreamJob. В таком случае сделаем отдельный мердж с помощью right join чтобы не потерять данные компании

```python
df_content = c.merge(
    v,
    on="company_name",
    how="right",
    suffixes=("_company", "_video")
)

df_content.head()
```

- общий смердженный для контента (right join)
