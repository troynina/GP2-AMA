```python
dict_rus_eng = {
    'Astrum Entertainment': 'аструм',
    'ИТ-холдинг Т1': 'т1',
    '1520 Сигнал (входит в экосистему РЖД)': 'ржд',
    'Лаборатория Касперского': 'касперский',
    'Ростелеком (цифровые активы)': 'ростелеком',
    'Сбер (цифровые активы)': 'сбер',
    'МТС (цифровые активы)': 'мтс',
    'Т-Банк': 'тбанк',
    'Yandex': 'яндекс',
    'Wildberries': 'вб',
    'Ozon': 'озон',
    'Edna': 'эдна',
    'Softline': 'софтлайн',
    'VK': 'вк',
    'Lamoda': 'ламода',
    'HH': 'хх',
    'Saby (Тензор)': 'тензор',
    'CloudPayments + CloudTips + CloudKassir': 'cloud',
    'Naumen': 'наумен',
    'VisorLabs': 'визорлабс',
    'Arenadata': 'аренадата',
    'Холдинг Performance Group': 'performance group',
    'BelkaCar': 'белка кар',
    'Getcourse': 'геткурс',
    'Grow Food': 'гроу фуд',
    'Litres (принадлежит Эксмо-АСТ)': 'литрес'
}
```

```python
import re

def normalize_name(name):
  name = name.strip()
  name = re.sub(r'^гк\s+', '', name, flags=re.IGNORECASE)  # убрать ГК в начале - это влияет на запросы в вордстате
  name = re.sub(r'\s*\(цифровые активы\)\s*$', '', name, flags=re.IGNORECASE) #убираем в конце цифровые активы чтобы не портить статистику запросов
  return name.strip()

def get_value(name):
  cleaned_name = normalize_name(name)

  for k, v in dict_rus_eng.items():
    if normalize_name(k).lower() == cleaned_name.lower():
      return v
  return cleaned_name.lower()

bigtech['search_name'] = bigtech['company'].apply(get_value)
```

```python
import time
import requests

demand = [' вакансии', ' стажировка', ' собеседования', ' зп', ' зарплата'] #запросы которые будем испоьзовать для каждой компании

results = []

for name in bigtech['search_name']:
  row = {'company': name}
  total = 0

  for d in demand:
    phrase = name + d
    params = {
      'phrase': phrase,
      'devices': ['DEVICE_ALL'],
      'folderId': FOLDER_ID,
    }
    resp = requests.post(url, headers=headers, json=params, timeout=30)

    print(f"Запрос: {phrase}, статус: {resp.status_code}")
    if resp.status_code != 200:
        print(f"Ошибка: {resp.text}")
        continue
    data = resp.json()

    top_requests = data.get('results', [])
    q_count = sum(int(result['count']) for result in top_requests)

    row[d.strip()] = q_count
    row[d.strip() + '_top'] = top_requests

    total += q_count
    time.sleep(0.2)

  row['total'] = total
  results.append(row)

df = pd.DataFrame(results)
bdf = bigtech.copy()
df

```

```python
merged = pd.merge(bdf, df, left_on='search_name', right_on='company', how='left', suffixes=('', '_df'))
merged.drop(columns='company_df', inplace=True)

merged[['вакансии', 'стажировка', 'собеседования', 'зп', 'зарплата', 'total']] = merged[['вакансии', 'стажировка', 'собеседования', 'зп', 'зарплата', 'total']].fillna(0).astype(int)

merged.head()
```

Ланит судя по всему не прошел до конца обработку названия - дополним вручную информацию по нему чтобы не тратить токены запросов прогоняя весь алгоритм заново

```python
lanit_name = '«ланит»:'
lanit_query = 'ланит'

row = {'company': lanit_name}
total = 0

for d in demand:
  phrase = lanit_query + d
  params = {
      'phrase': phrase,
      'devices': ['DEVICE_ALL'],
      'folderId': FOLDER_ID,
  }

  resp = requests.post(url, headers=headers, json=params, timeout=30)

  print(f'Запрос: {phrase}, статус: {resp.status_code}')
  if resp.status_code != 200:
    print(f'Ошибка: {resp.text}')
    row[d.strip()] = 0
    row[d.strip() + '_top'] = []
    continue

  data = resp.json()
  top_requests = data.get('results', [])
  q_count = sum(int(result['count']) for result in top_requests)

  row[d.strip()] = q_count
  row[d.strip() + '_top'] = top_requests
  total += q_count

  time.sleep(0.2)

row['total'] = total

lanit_df = pd.DataFrame([row])
lanit_df
```

```python
lanit_df['зарплата_общая'] = lanit_df['зп'] + lanit_df['зарплата']
lanit_df['зарплата_top_общий'] = lanit_df['зп_top'] + lanit_df['зарплата_top']

lanit_df.drop(columns=['зп', 'зп_top', 'зарплата', 'зарплата_top'], inplace=True)

lanit_df
```
